```python
import streamlit as st
import time

# -------------------------------
# Title
# -------------------------------
st.title("🇮🇳 BharatFilter")
st.subheader("Multilingual Toxic Content Detector")

# -------------------------------
# Toxic Words Dictionary
# -------------------------------
toxic_words = [
    "hate", "kill", "stupid", "idiot",
    "बेवकूफ", "नफरत", "मार",
    "मूर्ख", "द्वेष", "मार"
]

# -------------------------------
# Naive Algorithm
# -------------------------------
def naive_search(text, pattern):
    n = len(text)
    m = len(pattern)
    positions = []

    for i in range(n - m + 1):
        match = True
        for j in range(m):
            if text[i + j] != pattern[j]:
                match = False
                break
        if match:
            positions.append(i)

    return positions

# -------------------------------
# Rabin-Karp Algorithm
# -------------------------------
def rabin_karp(text, pattern, d=256, q=101):
    n = len(text)
    m = len(pattern)
    positions = []

    if m > n:
        return positions

    p = 0
    t = 0
    h = 1

    for i in range(m - 1):
        h = (h * d) % q

    for i in range(m):
        p = (d * p + ord(pattern[i])) % q
        t = (d * t + ord(text[i])) % q

    for i in range(n - m + 1):
        if p == t:
            if text[i:i+m] == pattern:
                positions.append(i)

        if i < n - m:
            t = (d * (t - ord(text[i]) * h) + ord(text[i + m])) % q
            if t < 0:
                t += q

    return positions

# -------------------------------
# Input Box
# -------------------------------
user_input = st.text_area("Enter Social Media Text")

# -------------------------------
# Detect Button
# -------------------------------
if st.button("Detect Toxic Content"):

    if user_input.strip() == "":
        st.warning("Please enter some text!")
    else:
        text = user_input.lower()

        # NAIVE
        start_naive = time.time()
        naive_results = {}
        for word in toxic_words:
            pos = naive_search(text, word.lower())
            if pos:
                naive_results[word] = pos
        end_naive = time.time()

        # RABIN-KARP
        start_rk = time.time()
        rk_results = {}
        for word in toxic_words:
            pos = rabin_karp(text, word.lower())
            if pos:
                rk_results[word] = pos
        end_rk = time.time()

        # Final Status
        detected = bool(naive_results or rk_results)

        st.subheader("🔍 Final Result")
        if detected:
            st.error("⚠️ Toxic Content Detected")
        else:
            st.success("✅ No Toxic Content Detected")

        # Naive Output
        st.subheader("Naive Algorithm")
        if naive_results:
            for word, pos in naive_results.items():
                st.write(f"'{word}' found at positions {pos}")
        else:
            st.write("No toxic words found")

        st.write(f"Time: {end_naive - start_naive:.6f} sec")

        # Rabin-Karp Output
        st.subheader("Rabin-Karp Algorithm")
        if rk_results:
            for word, pos in rk_results.items():
                st.write(f"'{word}' found at positions {pos}")
        else:
            st.write("No toxic words found")

        st.write(f"Time: {end_rk - start_rk:.6f} sec")
```
